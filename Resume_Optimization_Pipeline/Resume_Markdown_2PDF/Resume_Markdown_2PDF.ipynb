```python
# ── 2. DEFINE PDF STYLES ──────────────────────────────────────────────────────
def build_styles():
    base  = getSampleStyleSheet()
    DARK  = colors.HexColor('#1A1A2E')
    BLUE  = colors.HexColor('#0F3460')
    ACCENT= colors.HexColor('#16213E')
    LIGHT = colors.HexColor('#E8E8E8')

    styles = {
        'name': ParagraphStyle(
            'name', fontName='Helvetica-Bold',
            fontSize=20, textColor=DARK,
            spaceAfter=2, alignment=TA_LEFT
        ),
        'section': ParagraphStyle(
            'section', fontName='Helvetica-Bold',
            fontSize=11, textColor=BLUE,
            spaceBefore=10, spaceAfter=4
        ),
        'subsection': ParagraphStyle(
            'subsection', fontName='Helvetica-Bold',
            fontSize=10, textColor=DARK,
            spaceBefore=6, spaceAfter=2
        ),
        'body': ParagraphStyle(
            'body', fontName='Helvetica',
            fontSize=9.5, textColor=colors.HexColor('#333333'),
            spaceAfter=3, leading=14
        ),
        'bullet': ParagraphStyle(
            'bullet', fontName='Helvetica',
            fontSize=9.5, textColor=colors.HexColor('#333333'),
            leftIndent=15, spaceAfter=2,
            leading=13, bulletIndent=5
        ),
    }
    return styles
```

```python
# ── 3. MARKDOWN → PDF CONVERTER ───────────────────────────────────────────────
def markdown_to_pdf(md_text, output_path, role):
    """
    Converts Markdown resume text to a professional PDF.
    Handles: # H1, ## H2, ### H3, - bullets, plain text
    Outputs ATS-compatible, multi-platform PDF.
    """
    doc = SimpleDocTemplate(
        output_path,
        pagesize=letter,
        rightMargin=0.65*inch, leftMargin=0.65*inch,
        topMargin=0.65*inch,   bottomMargin=0.65*inch
    )

    styles = build_styles()
    story  = []

    # Header color bar (visual polish)
    story.append(Table(
        [['']],
        colWidths=[7.2*inch],
        rowHeights=[6],
        style=TableStyle([
            ('BACKGROUND', (0,0), (-1,-1), colors.HexColor('#0F3460')),
            ('LINEBELOW',  (0,0), (-1,-1), 0, colors.white),
        ])
    ))
    story.append(Spacer(1, 6))

    lines = md_text.strip().split('\n')

    for line in lines:
        line = line.strip()
        if not line:
            story.append(Spacer(1, 4))
            continue

        # H1 — Candidate Name
        if line.startswith('# '):
            text = line[2:].strip()
            story.append(Paragraph(text, styles['name']))
            story.append(HRFlowable(
                width="100%", thickness=1.5,
                color=colors.HexColor('#0F3460'), spaceAfter=6
            ))

        # H2 — Section Header
        elif line.startswith('## '):
            text = line[3:].strip().upper()
            story.append(Spacer(1, 4))
            story.append(Paragraph(text, styles['section']))
            story.append(HRFlowable(
                width="100%", thickness=0.5,
                color=colors.HexColor('#CCCCCC'), spaceAfter=3
            ))

        # H3 — Job Title / Subsection
        elif line.startswith('### '):
            text = line[4:].strip()
            story.append(Paragraph(text, styles['subsection']))

        # Bullet points
        elif line.startswith('- '):
            text = line[2:].strip()
            # bold key: value pattern
            text = re.sub(r'\*\*(.*?)\*\*', r'<b>\1</b>', text)
            story.append(Paragraph(f'• {text}', styles['bullet']))

        # Plain text
        else:
            text = re.sub(r'\*\*(.*?)\*\*', r'<b>\1</b>', line)
            story.append(Paragraph(text, styles['body']))

    # Footer
    story.append(Spacer(1, 12))
    story.append(HRFlowable(width="100%", thickness=0.5,
                             color=colors.HexColor('#CCCCCC')))
    story.append(Paragraph(
        f'<font size=7 color="#888888">Generated by AI Resume Rewriting System '
        f'| Role: {role} | Optimized for ATS compatibility</font>',
        ParagraphStyle('footer', alignment=TA_CENTER, spaceAfter=0)
    ))

    doc.build(story)
```

```python
# ── 4. CONVERT ALL RESUMES TO PDF ─────────────────────────────────────────────
print("🔄 Converting Markdown Resumes to PDF...\n")
pdf_files = []

for role, md_text in rewritten_resumes.items():
    filename = f"resume_{role.replace('-', '_').lower()}.pdf"
    markdown_to_pdf(md_text, filename, role)
    pdf_files.append(filename)
    print(f"   ✅ PDF Created → {filename}")

print(f"\n✅ {len(pdf_files)} PDF resumes generated successfully!")
```

```python
# ── 5. VERIFY PDF FILES ───────────────────────────────────────────────────────
print("\n📁 PDF File Sizes:\n")
for f in pdf_files:
    size = os.path.getsize(f)
    print(f"   {f:<45} → {size/1024:.1f} KB")
```

```python
# ── 6. DOWNLOAD ALL PDFs ──────────────────────────────────────────────────────
print("\n📥 Downloading PDFs to local system...\n")
from google.colab import files
for f in pdf_files:
    files.download(f)
    print(f"   ⬇️  Downloading: {f}")

print("\n" + "="*60)
print("   Markdown resumes converted to professional PDFs")
print("   ATS-compatible format for multi-platform use")
print("   All PDFs downloaded to your local system")
print("="*60)
```
